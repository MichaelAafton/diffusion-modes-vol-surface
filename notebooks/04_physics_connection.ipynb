## Phase 4: Connecting PCA to the Heat Equation

This is the notebook that makes the project unique. We show quantitatively
that the empirical PCA results from Notebook 03 match the theoretical
predictions of the stochastic heat equation.

**Key tests**:
1. Eigenvalue scaling: $\lambda_k \sim k^{-\alpha}$ with $\alpha \approx 2$
2. Eigenmode shape: PCA modes ≈ Fourier modes (cosines)
3. Correlation structure: empirical vs analytical Green's function
4. Fit the diffusion coefficient $D$ from the correlation matrix

```python
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.insert(0, '..')

from src.pca import run_pca
from src.heat_equation import (
    fourier_eigenfunctions, theoretical_eigenvalue_spectrum,
    fit_power_law, compare_modes, fit_diffusion_coefficient,
)
from src.plotting import apply_style, plot_eigenmodes, plot_eigenvalue_spectrum

apply_style()

# Load data
data = np.load('../data/synthetic/spde_dataset.npz')
z_grid = data['z_grid']
returns = data['returns']
D_true = float(data['D'])
print(f'True diffusion coefficient: D = {D_true}')
```

```python
# Run PCA
pca_result = run_pca(returns, n_components=15)
```

## 4.1 — Eigenvalue scaling: is α ≈ 2?

```python
# Fit power law to eigenvalues
power_law = fit_power_law(pca_result.eigenvalues[:10])  # first 10 modes

print(f"Fitted exponent: α = {power_law['alpha']:.3f}  (theory predicts α = 2.0)")
print(f"R² = {power_law['r_squared']:.4f}")

# TODO: Plot the log-log eigenvalue spectrum with the fitted line
# and the theoretical k^-2 reference
```

## 4.2 — Mode comparison: PCA vs Fourier

```python
# Compare empirical modes to theoretical cosines
comparison = compare_modes(pca_result.eigenvectors[:5], z_grid)

print('Mode overlaps (1.0 = perfect match):')
for k, overlap in enumerate(comparison['overlaps']):
    print(f'  Mode {k}: {overlap:.4f}')
print(f'  Mean overlap: {comparison["mean_overlap"]:.4f}')

# Plot eigenmodes with theoretical overlay
fig, axes = plot_eigenmodes(
    pca_result, z_grid, n_modes=5,
    theoretical_modes=comparison['theoretical_modes'],
    title='Empirical PCA Modes vs Fourier Eigenfunctions',
)
plt.show()
```

## 4.3 — Fit the diffusion coefficient

```python
# Fit D from the empirical correlation matrix
fit_result = fit_diffusion_coefficient(pca_result.correlation_matrix, z_grid)

print(f"Fitted D = {fit_result['D']:.4f}  (true D = {D_true})")
print(f"Relative error in D: {abs(fit_result['D'] - D_true)/D_true*100:.1f}%")
print(f"Correlation matrix relative error: {fit_result['relative_error']*100:.1f}%")

# TODO: Plot empirical vs model correlation matrix side by side
```
