# 04 — The Physics Connection (where it deviates is the finding)

**Phase 4.** Connect the empirical PCA to the stochastic heat equation and quantify
the match honestly. On the synthetic harness everything matches by construction; on
real data the **deviation** from the heat-equation prediction is the contribution.

```python
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.insert(0, '..')

from src.pca import run_pca
from src.heat_equation import fit_power_law, compare_modes, fit_diffusion_coefficient
from src.plotting import apply_style, plot_eigenmodes
apply_style()

data = np.load('../data/synthetic/spde_dataset.npz')
z_grid, returns, D_true = data['z_grid'], data['returns'], float(data['D'])
pca = run_pca(returns, n_components=15)
print(f'true diffusion constant D = {D_true}')
```

## 4.1 — Eigenvalue scaling (the central real-data test)

Fit $\lambda_k \sim k^{-\alpha}$ on log–log axes. The heat equation predicts $\alpha = 2$.
On synthetic data $\alpha \approx 2$ **by construction**; on real data, *where the
spectrum departs from $k^{-2}$ is where the market is not a clean diffusive membrane*.

```python
pl = fit_power_law(pca.eigenvalues[:10])
print(f"alpha = {pl['alpha']:.3f}  (heat equation predicts 2.0),  R^2 = {pl['r_squared']:.4f}")
```

## 4.2 — Empirical modes vs Fourier (Laplacian) eigenfunctions

```python
cmp = compare_modes(pca.eigenvectors[:5], z_grid)
for k, o in enumerate(cmp['overlaps']):
    print(f'  mode {k}: overlap {o:.4f}')
plot_eigenmodes(pca, z_grid, n_modes=5, theoretical_modes=cmp['theoretical_modes'],
                title='Empirical PCA modes vs Fourier eigenfunctions'); plt.show()
```

## 4.3 — Recover the diffusion constant (harness check)

```python
fit = fit_diffusion_coefficient(pca.correlation_matrix, z_grid)
print(f"fitted D = {fit['D']:.4f}  (true D = {D_true};  rel. error "
      f"{abs(fit['D']-D_true)/D_true*100:.1f}%)")
```
