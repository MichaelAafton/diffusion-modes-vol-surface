# 03 — PCA Analysis

## Phase 3: Discovering eigenmodes of the volatility surface

This is the core analysis notebook. We run PCA on the simulated
option returns to extract the dominant modes of variation.

**Expected results (Slides 6-8)**:
- Eigenmode 0: ~flat (parallel shift) — ~84% of variance
- Eigenmode 1: ~linear (tilt/skew) — ~11.5%
- Eigenmode 2: ~quadratic (smile) — ~3%
- Higher modes: increasingly oscillatory
- Eigenvalues decay as ~k⁻²
- Modes are stable across rolling windows

```python
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.insert(0, '..')

from src.pca import run_pca, rolling_pca
from src.plotting import (
    apply_style, plot_eigenmodes, plot_eigenvalue_spectrum,
    plot_correlation_matrix, plot_rolling_stability,
)

apply_style()

# Load the synthetic dataset from Notebook 01
data = np.load('../data/synthetic/spde_dataset.npz')
z_grid = data['z_grid']
returns = data['returns']
print(f'Loaded: {returns.shape[0]} days, {returns.shape[1]} grid points')
```

## 3.1 — PCA on full sample

```python
# Run PCA
pca_result = run_pca(returns, n_components=10)

print('Explained variance ratios:')
for i, ev in enumerate(pca_result.explained_variance_ratio):
    print(f'  Mode {i}: {ev*100:.1f}%')
print(f'  Total (first 10): {pca_result.explained_variance_ratio.sum()*100:.1f}%')
```

```python
# Plot eigenmodes (reproduce Slide 6)
fig, axes = plot_eigenmodes(pca_result, z_grid, n_modes=5)
plt.show()
```

```python
# Eigenvalue spectrum — bar chart + log-log (Slides 6 and 9)
fig, axes = plot_eigenvalue_spectrum(pca_result)
plt.show()
```

```python
# Correlation matrix
fig, ax = plot_correlation_matrix(pca_result.correlation_matrix, z_grid)
plt.show()
```

## 3.2 — Rolling PCA: stability over time (Slide 8)

Run PCA on 1-year rolling windows to verify that the eigenmodes
are stable — a key finding of the presentation.

```python
# Rolling PCA
rolling_results = rolling_pca(returns, window_size=250, step_size=50, n_components=5)
print(f'Number of rolling windows: {len(rolling_results)}')

# Plot stability
fig, axes = plot_rolling_stability(rolling_results, z_grid, n_modes=5)
plt.show()
```
