# 03 — PCA Analysis: On Both, Compared

**Phase 3 (core empirical work).** Run *identical* PCA on the synthetic harness and on
real data, and compare — the contrast is the point.

Expected shape: mode 0 ~flat (parallel shift, bulk of variance), mode 1 ~monotone
(skew/tilt), mode 2 ~one-node (smile), higher modes increasingly oscillatory. On
synthetic data this is clean by construction. **On real data, report the actual
variance shares and how much they differ — that difference is data, not error.**

```python
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.insert(0, '..')

from src.pca import run_pca, rolling_pca
from src.plotting import (apply_style, plot_eigenmodes, plot_eigenvalue_spectrum,
                          plot_correlation_matrix, plot_rolling_stability)
apply_style()

data = np.load('../data/synthetic/spde_dataset.npz')
z_grid, returns = data['z_grid'], data['returns']
print(f'synthetic harness: {returns.shape[0]} days x {returns.shape[1]} grid points')
```

## 3.1 — PCA on the synthetic harness (recover ground truth)

```python
pca = run_pca(returns, n_components=10)
for i, ev in enumerate(pca.explained_variance_ratio):
    print(f'  mode {i}: {ev*100:5.1f}%')
```

```python
plot_eigenmodes(pca, z_grid, n_modes=5); plt.show()
plot_eigenvalue_spectrum(pca); plt.show()
plot_correlation_matrix(pca.correlation_matrix, z_grid); plt.show()
```

## 3.2 — Rolling-window stability

```python
rolling = rolling_pca(returns, window_size=250, step_size=50, n_components=5)
print(f'{len(rolling)} rolling windows')
plot_rolling_stability(rolling, z_grid, n_modes=5); plt.show()
```
